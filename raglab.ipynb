### 🦜 Iteración 1: Construyendo un "loro"

**¿Qué construimos aquí?**

En esta etapa inicial, implementamos un servidor Flask básico que simula un chatbot. Este "dummy bot" no utiliza ningún modelo de lenguaje ni lógica inteligente, simplemente repite o transforma el mensaje recibido.

**¿Por qué hacerlo así primero?**

- ✅ Nos aseguramos de que el frontend y el backend pueden comunicarse correctamente.
- ✅ Probamos la estructura de peticiones (`/ask_bot`) y la respuesta en formato JSON.
- ✅ Es el primer paso hacia una arquitectura modular donde luego reemplazaremos el "dummy" por un motor LLM real.

Este paso es fundamental para validar la conexión entre el cliente (HTML/JS) y el servidor Flask, antes de añadir la complejidad de LangChain o RAG.


```python
import threading
import time
from flask import Flask, render_template_string, request, jsonify

stop_flag = False

app = Flask(__name__)

@app.route('/')
def home():
    return render_template_string(html)

# Placeholder para el chatbot
@app.route('/ask_bot', methods=['POST'])
def ask_bot():
    data = request.form
    return jsonify({'msg': f"🦜 dice: {data.get('msg', '')}"})

def run_server():
    from werkzeug.serving import make_server
    server = make_server('0.0.0.0', 5002, app)
    server.timeout = 1  # para que revise la bandera de salida cada segundo
    while not stop_flag:
        server.handle_request()
    print("Servidor detenido.")

# Guardamos el hilo
server_thread = threading.Thread(target=run_server)
server_thread.start()
!npx localtunnel --port 5002
```

### 🧠 Paso 2: Chatbot Básico con Memoria

**¿Qué hacemos en este paso?**

Aquí construimos un chatbot utilizando un modelo de lenguaje real (LLM) y le añadimos memoria conversacional. Esto permite al bot mantener el contexto de la conversación entre turnos, generando respuestas más naturales y coherentes.

**Tecnología clave:** `ConversationBufferMemory` de LangChain

**¿Por qué es importante la memoria en marketing conversacional?**

- 🔁 Mejora la experiencia del usuario al permitir referencias a mensajes anteriores.
- 🎯 Facilita diálogos naturales como: *"¿Y qué promociones tengo si ya soy cliente?"*
- 🧩 Permite construir asistentes que simulan atención al cliente de manera fluida.

Este paso transforma el dummy bot inicial en un asistente conversacional más realista, base esencial antes de incorporar RAG o herramientas externas.


```python
import os
from langchain_openai import ChatOpenAI
from langchain_core.runnables.history import RunnableWithMessageHistory
from langchain_core.chat_history import (
    BaseChatMessageHistory,
    InMemoryChatMessageHistory,
)
from langchain_core.messages import SystemMessage, HumanMessage

model = ChatOpenAI(model="gpt-4o-mini")

prompt = "Eres un chatbot técnico que ayuda a estudiantes a crear chatbots basados en Langchain y OpenAI. Tu nombre es YiPiti."

store = {}
def get_session_history(session_id: str) -> BaseChatMessageHistory:
    # RunnableWithMessageHistory llama a esta función solo con session_id (sin `self`)
    if session_id not in store:
        store[session_id] = InMemoryChatMessageHistory()
    return store[session_id]

config = {"configurable": {"session_id": "test"}}
chat = RunnableWithMessageHistory(model, get_session_history)
chat.invoke([SystemMessage(content=prompt)], config=config)
```

```python
from flask import Flask, request, jsonify
from werkzeug.serving import make_server
import threading

# Reinicia todo el entorno de servidor
try:
    stop_flag = True
    server_thread.join()
except:
    pass  # si no está definido aún

# Re-crear app limpia
app = Flask(__name__)

# Si ya estaba definida, eliminarla
if 'ask_bot' in app.view_functions:
    del app.view_functions['ask_bot']

@app.route('/')
def home():
    return render_template_string(html)

@app.route('/ask_bot', methods=['POST'])
def ask_bot():
    data = request.form or request.json
    msg, session_id, msg_id = data['msg'], data['session_id'], data['msg_id']
    bot_response = chat.invoke([HumanMessage(content=msg)], config=config).content
    response = {'msg': bot_response, 'session_id': session_id}
    return response

# Bandera de control
stop_flag = False

# Lanzar el servidor en un hilo
server_thread = threading.Thread(target=run_server)
server_thread.start()
!npx localtunnel --port 5002
```

### 📚 Paso 3: Chatbot con RAG (Retrieval-Augmented Generation)

**¿Qué implementamos en este paso?**

Aquí transformamos el chatbot para que pueda consultar una base de conocimiento antes de generar respuestas. Esto se conoce como **RAG: Retrieval-Augmented Generation**. En lugar de depender únicamente del conocimiento del modelo de lenguaje, el chatbot busca información relevante en una colección de documentos previamente cargados.

**Tecnologías utilizadas:**
- `FAISS` (o `Chroma`) para almacenar y recuperar documentos vectorizados
- `ConversationalRetrievalChain` para integrar recuperación + generación
- `OpenAIEmbeddings` para representar semánticamente los textos

**¿Por qué RAG es útil?**

- 📊 Permite responder preguntas basadas en información específica de productos, campañas o estudios internos.
- 🎯 Da respuestas actualizadas sin necesidad de reentrenar el modelo.
- 📎 Facilita el acceso a documentos institucionales o catálogos durante una conversación.

Con RAG, el chatbot se vuelve no solo conversacional, sino también **informado**. Esto es clave en escenarios reales como atención al cliente, soporte técnico o asistentes de ventas.


```python
from langchain_community.vectorstores import FAISS
from langchain_openai import OpenAIEmbeddings
from langchain_text_splitters import RecursiveCharacterTextSplitter

# Datos de ejemplo para alimentar nuestro retrieval
docs = [
    "LangChain permite construir RAG componiendo loaders, splitters, embeddings, vector stores y LLMs.",
    "Para PDFs usa PyPDFLoader; para Markdown usa UnstructuredMarkdownLoader; conserva metadatos (ruta, página).",
    "Chunking recomendado: 800–1200 caracteres con solape 100–200 usando RecursiveCharacterTextSplitter.",
    "Embeddings sugeridos: BAAI/bge-m3 (multilingüe) o sentence-transformers/all-MiniLM-L6-v2 (ligero).",
    "FAISS (IndexFlatIP) es práctico en notebooks; normaliza los embeddings para similitud coseno.",
    "Convierte FAISS a retriever con search_kwargs={'k': 4}; ajusta k entre 3–8 según resultados.",
    "Formatea el contexto con metadatos y exige en el prompt respuestas breves y sección 'Fuentes:'.",
    "LCEL: {'context': retriever|format, 'question': passthrough} | prompt | llm | StrOutputParser.",
    "La memoria conversacional (RunnableWithMessageHistory) no sustituye al RAG; la evidencia del retriever manda.",
    "Activa streaming de tokens para mejor UX y limita max_new_tokens a 256–512 en respuestas técnicas.",
    "Evalúa con preguntas reales del equipo: groundedness (citas), faithfulness (sin inventos) y latencia.",
    "Versiona embeddings/índices; en FAISS usa save_local()/load_local() para reproducibilidad.",
    "Usa filtros por metadatos (archivo, sección) para mejorar precisión en colecciones grandes.",
    "Combina BM25 + embeddings (búsqueda híbrida) para consultas con términos exactos.",
    "Temperatura recomendada 0.2–0.4 para respuestas más deterministas en documentación técnica."
]

# Divide el texto en chunks para indexación
splitter = RecursiveCharacterTextSplitter(chunk_size=100, chunk_overlap=20)
split_docs = splitter.create_documents(docs)

# Genera embeddings y construye el índice vectorial (FAISS)
embedding = OpenAIEmbeddings(model="text-embedding-3-small")

db = FAISS.from_documents(split_docs, embedding)
```

### 🤖 Paso 4: Chatbot con ReAct (Razonamiento + Acción con herramientas)

**¿Qué hacemos en este paso?**

Integramos el agente ReAct, una técnica avanzada que permite al chatbot **razonar paso a paso y ejecutar acciones** usando herramientas externas, como una búsqueda en Wikipedia o una consulta en la base de conocimiento.

**¿Qué significa ReAct?**  
*ReAct* combina **Reasoning** (razonamiento) + **Acting** (acciones usando herramientas) para generar respuestas más completas y contextualizadas.

**Herramientas que usamos:**
- 🔍 Wikipedia: para obtener conocimiento general
- 🧠 RAG (FAISS): para conocimientos específicos de marketing

**¿Por qué ReAct es útil ?**

- Permite responder a preguntas más amplias que requieren datos externos + internos.
- Simula cómo actuaría un humano: busca, consulta fuentes y responde.
- Abre la puerta a asistentes inteligentes que usan múltiples sistemas (CRM, web, documentos).

Este paso eleva el chatbot a un nuevo nivel, convirtiéndolo en un **asistente autónomo capaz de tomar decisiones** sobre qué información usar para dar la mejor respuesta.


```python
from langchain_openai import ChatOpenAI
from langchain.agents import create_agent
from langchain_community.utilities import WikipediaAPIWrapper
from langchain_core.tools import Tool

wiki = WikipediaAPIWrapper()
wiki_tool = Tool(
    name="Wikipedia",
    func=wiki.run,
    description="Úsalo para buscar información general en Wikipedia."
)

def run_rag(q: str) -> str:
    result = qa_chain.invoke(q) if hasattr(qa_chain, "invoke") else qa_chain.run(q)
    return result["result"] if isinstance(result, dict) else result

rag_tool = Tool(
    name="RAGDocsTecnicos",
    func=run_rag,
    description="Usa esta herramienta por defecto, y asume que la mayoria de preguntas son de documentacion tecnica. Úsalo para responder preguntas específicas basadas en la documentación técnica indexada."
)

tools = [wiki_tool, rag_tool]

llm = ChatOpenAI(model="gpt-4.1-mini", temperature=0.8)

agent = create_agent(
    model=llm,
    tools=tools,
    system_prompt="Eres un asistente que decide cuándo usar Wikipedia o la documentación técnica indexada para responder."
)

if __name__ == "__main__":
    pregunta = "Qué temperatura funciona para hacer documentación?"
    respuesta = agent.invoke({"messages": [{"role": "user", "content": pregunta}]})
    print("Respuesta final del agente:", respuesta["messages"][-1].content)
```
